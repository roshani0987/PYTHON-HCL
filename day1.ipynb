```python
x=5
y=3
z=7
#comparison operators
print(x==y) #False
print(z==y) #False
print(x!=y) #True
print(x>y) #True
print(x<=z) #True
#logical operators
print(not(z>y)) #False
print(x>y and x<z) #True
print(x>y or x<z) #True
```

```python
#Function is a block of reusable code only executes when it is called
#lamda is a small anonymous function that can take any number of arguments, but can only have one expression. It is defined using the `lambda` keyword.
#function with parameters
def add(x,y):
    return x + y
add(5, 10)  # This will return 15

#function without parameters
def add():
    x,y = 10,20
    return x + y
add()  # This will return 30
```
